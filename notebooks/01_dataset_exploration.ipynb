```python
import pandas as pd
from scipy.io import arff

file_path = "../data/raw/phishing_websites.arff"

data, meta = arff.loadarff(file_path)

df = pd.DataFrame(data)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# Convert byte values to normal integers

for column in df.columns:
    df[column] = df[column].apply(
        lambda x: int(x.decode()) if isinstance(x, bytes) else x
    )

print(df.head())
```

```python
# Check whether duplicate feature combinations have different labels

feature_columns = df.columns.drop("Result")

duplicate_groups = (
    df[df.duplicated(subset=feature_columns, keep=False)]
    .groupby(list(feature_columns))["Result"]
    .nunique()
)

conflicting_duplicates = (duplicate_groups > 1).sum()

print("Duplicate feature groups with conflicting labels:",
      conflicting_duplicates)
```

```python
# Remove exact duplicate rows

df = df.drop_duplicates().reset_index(drop=True)

print("Dataset shape after removing duplicates:", df.shape)
```

```python
df.to_csv(
    "../data/processed/phishing_websites_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```

```python
# Separate features and target

X = df.drop("Result", axis=1)
y = df["Result"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Convert target labels

y = y.map({
    -1: 0,
     1: 1
})

print(y.value_counts())
```

```python
print("Unique target values:", y.unique())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
print("Training class distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting class distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
from sklearn.ensemble import RandomForestClassifier

# Create the model
rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

# Train the model
rf_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

```python
# Make predictions on test data

y_pred = rf_model.predict(X_test)

print("Predictions generated successfully!")
print(y_pred[:20])
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Legitimate", "Phishing"]
))
```

```python
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
import joblib

model_path = "../models/random_forest_url_model.joblib"

joblib.dump(rf_model, model_path)

print("Model saved successfully!")
print("Location:", model_path)
```

```python
# Get feature importance

feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print(feature_importance)
```

```python
top_features = feature_importance.head(10)

print(top_features)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.barh(
    top_features["Feature"][::-1],
    top_features["Importance"][::-1]
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Top 10 Features Used by Random Forest")

plt.show()
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    display_labels=["Legitimate", "Phishing"]
)

plt.title("PhishGuard - Confusion Matrix")
plt.show()
```
